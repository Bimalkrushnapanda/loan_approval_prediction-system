```python
import pandas as pd
```

```python
df=pd.read_csv("loan_approval_data.csv")
```

```python
df.sample(2)
```

```python
df.isna().sum()
```

```python
no_nan=df.dropna()
```

```python
no_nan
```

```python
df.drop(columns=["Applicant_ID"],inplace=True)
```

```python
df
```

```python
df=pd.get_dummies(df,columns=["Employment_Status","Marital_Status","Loan_Purpose","Property_Area","Education_Level","Gender","Employer_Category"],drop_first=True,dtype=int)
```

```python
df
```

```python
for i in ["Applicant_Income","Coapplicant_Income","Age","Savings","Collateral_Value","Loan_Amount","Loan_Term"]:
    df[i]=df[i].fillna(df[i].mean())
```

```python
for i in ["Dependents","Credit_Score","Existing_Loans","DTI_Ratio"]:
    df[i]=df[i].fillna(0)
```

```python
df
```

```python
df.isna().sum()
```

```python
df=df.dropna()
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
from sklearn import metrics
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
```

```python
Y=df["Loan_Approved"]
X=df.drop(columns=["Loan_Approved"])
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=.25,random_state=42)
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
pipe=Pipeline([
    ("sc",StandardScaler()),
    ("knn",KNeighborsClassifier())
])
param_grid={"knn__n_neighbors":[3,5,7,9]}
classifier=GridSearchCV(
    pipe,
    param_grid,
    cv=5
)
```

```python
classifier.fit(X_train,Y_train)
```

```python
Y_pred=classifier.predict(X_test)
```

```python
print(metrics.accuracy_score(Y_test,Y_pred))
```

```python
#for logistic regression
pipe=Pipeline([
    ("sc",StandardScaler()),
    ("model",LogisticRegression())
])
```

```python
pipe.fit(X_train,Y_train)
```

```python
Y_pred=pipe.predict(X_test)
```

```python
print(metrics.accuracy_score(Y_test,Y_pred))
```
